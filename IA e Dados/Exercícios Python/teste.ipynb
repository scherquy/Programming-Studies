```python
import pandas as pd
```

```python
cidades = ["São Paulo (SP)",
           "Rio de Janeiro (RJ)",
           "Brasília (DF)",
           "Fortaleza (CE)",
           "Salvador (BA)",
           "Belo Horizonte (MG)",
           "Manaus (AM)",
           "Porto Alegre (RS)"
           ]

populacoes = [12,
              6.5,
              3,
              2.4,
              2.4,
              2.3,
              2,
              1.7
              ]
```

```python
cidades_s = pd.Series(cidades) #Cria a serie

cidades_s
```

```python
pop = pd.Series(populacoes) #Cria a serie

pop
```

```python
cidades_s[7] #Pega o valor da posição informada
```

```python
cidades_s[2:5] #Pega todos os valores dentro do range
```

```python
pop.max() #Pega o maior valor na serie
```

```python
pop.mean() #Tira a média dos elementos na serie
```

```python
pop * 2 #Multiplica a serie por um valor
pop / 2 #Divide a serie por um valor
pop + 2 #Soma a serie com um valor
pop - 2 #Subtrai a serie de um valor
```

```python
capitais = pd.DataFrame({"Cidade": cidades_s, "População": pop})

capitais
```
